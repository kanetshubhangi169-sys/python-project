```python
import pandas as pd
df = pd.read_csv("train.csv")
print(df.to_string())
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["Age"]=df["Age"].fillna(df["Age"].median())
df.head(9)
```

```python
df.isnull().sum()
```

```python
df["Embarked"]=df["Embarked"].fillna(df["Embarked"]).mode()[0]
df.head(835)
```

```python
df.isnull().sum()
```

```python
df.duplicated()
```

```python
df.shape
```

```python
survival_rate =(round(df["Survived"].mean() * 100 ,2))
print("survival rate:", survival_rate)
```

```python
gender_survival =(round(df.groupby("Sex") ["Survived"].mean()*100 ,2))
print(gender_survival)
```

```python
class_survival =(round(df.groupby("Pclass") ["Survived"].mean()*100 ,2))
print(class_survival)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)

plt.figure(figsize=(6,3))
sns.heatmap(corr,
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```
