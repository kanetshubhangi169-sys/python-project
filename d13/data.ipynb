```python
import pandas as pd
df = pd.read_csv("Students.csv")
print(df.shape)
print(df.columns)
```

```python
df['average'] = (df['math score'] + df['reading score'] + df['writing score'])/3
print(round(df['average'], 2).to_string())
```

```python
df.groupby('gender')['average'].mean()
```

```python
df.groupby('parental level of education')['average'].mean()
```

```python
top_students = (df.sort_values('average', ascending=False).head(10))
print(top_students.head(6))
```
